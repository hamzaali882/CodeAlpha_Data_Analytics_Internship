```python
# CodeAlpha Internship — Task 3: Data Visualization

## E-Commerce Book Dataset Visualization

#This project focuses on visualizing the book dataset collected during Task 1.
#The goal is to create clear and simple charts that help understand patterns in book prices, ratings, and categories.
```

```python
# Let's import the libraries we need for the visualizations

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Let's load the book dataset from Task 1

df = pd.read_csv("../task1_web_scraping/data/books_dataset.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Let's see how many books we have for each rating

rating_counts = df["Rating"].value_counts()

rating_counts.plot(
    kind="bar",
    figsize=(8, 5),
    title="Number of Books by Rating"
)

plt.xlabel("Rating")
plt.ylabel("Number of Books")
plt.xticks(rotation=0)
plt.show()
```

```python
# Let's check how the book prices are distributed

prices = df["Price_GBP"]

prices.plot(
    kind="hist",
    bins=15,
    figsize=(8, 5),
    title="Book Price Distribution"
)

plt.xlabel("Price (£)")
plt.ylabel("Number of Books")
plt.show()
```

```python
# Let's see which categories have the most books

top_categories = df["Category"].value_counts().head(10)

print(top_categories)
```

```python
# Let's make a chart for the top 10 categories

top_categories.plot(
    kind="bar",
    figsize=(10, 5),
    title="Top 10 Book Categories"
)

plt.xlabel("Category")
plt.ylabel("Number of Books")
plt.xticks(rotation=45)
plt.show()
```

```python
# Let's compare the average price of books for each rating

average_price_by_rating = df.groupby("Rating")["Price_GBP"].mean()

print(average_price_by_rating)
```

```python
# Let's make a chart to compare the average prices

average_price_by_rating.plot(
    kind="bar",
    figsize=(8, 5),
    title="Average Book Price by Rating"
)

plt.xlabel("Rating")
plt.ylabel("Average Price (£)")
plt.xticks(rotation=0)
plt.show()
```

```python
# Let's find the categories with the highest average prices

average_price_by_category = df.groupby("Category")["Price_GBP"].mean()

top_expensive_categories = average_price_by_category.sort_values(
    ascending=False
).head(10)

print(top_expensive_categories)
```

```python
# Let's make a chart for the categories with the highest average prices

top_expensive_categories.plot(
    kind="bar",
    figsize=(10, 5),
    title="Top 10 Categories by Average Price"
)

plt.xlabel("Category")
plt.ylabel("Average Price (£)")
plt.xticks(rotation=45)
plt.show()
```

```python
# Let's turn the ratings into numbers so we can compare them with price

rating_numbers = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

df["Rating_Number"] = df["Rating"].map(rating_numbers)

print(df[["Rating", "Rating_Number", "Price_GBP"]].head())
```

```python
# Let's see if book price changes with the rating

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Rating_Number"],
    df["Price_GBP"],
    alpha=0.5
)

plt.xlabel("Rating")
plt.ylabel("Price (£)")
plt.title("Book Price vs Rating")
plt.xticks([1, 2, 3, 4, 5])
plt.show()
```

```python
# Let's compare the price spread for each rating

df.boxplot(
    column="Price_GBP",
    by="Rating_Number",
    figsize=(8, 5)
)

plt.title("Price Distribution by Rating")
plt.suptitle("")
plt.xlabel("Rating")
plt.ylabel("Price (£)")
plt.show()
```

```python
# Let's make the category price comparison easier to read

top_expensive_categories.sort_values().plot(
    kind="barh",
    figsize=(10, 6),
    title="Top 10 Categories by Average Price"
)

plt.xlabel("Average Price (£)")
plt.ylabel("Category")
plt.show()
```

```python
## Key Findings

#- The dataset contains 1,000 books.
#- Most books have ratings between One and Five stars, with One-star books being the most common.
#- Book prices are spread across a range of £10.00 to £59.99.
#- Four-star books have the highest average price at around £36.09.
#- One-star books have the lowest average price at around £34.56.
#- There is no strong visible relationship between book rating and price.
#- Default and Nonfiction are among the categories with the highest number of books.
#- Suspense has the highest average price among the categories, at around £58.33.
```
